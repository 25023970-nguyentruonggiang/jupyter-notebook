W3A3
Kiểm tra xem một số có phải là lũy thừa của 2 hay không
Sử dụng các toán tử bitwise để xác định xem một số có phải là lũy thừa của 2 hay không.
Gợi ý: Đối với lũy thừa của 2, n & (n-1) == 0 


```python
x = int(input())
if x <= 0:
    print("Không phải luỹ thừa của 2")
else: 
    a=x-1
    if (x&a)== 0:
     print("Là luỹ thừa của 2")
    else:
     print("Không phải luỹ thừa của 2")
```

- 0-50 kWh: 1500đ/kWh
- 51-100 kWh: 2000đ/kWh
- 100 kWh: 3000đ/kWh 

```python
a=int(input())
if 0<=a<=50:
    tiendien=a*1500
    print(tiendien)
elif 51<=a<=100:
    tiendien=50*1500+(a-50)*2000
    print(tiendien)
else:
    tiendien=50*1500+50*2000+(a-100)*3000
    print(tiendien)
```
