Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python
a, b, c = map(float, input("Nhập 3 cạnh của tam giác: ").split())
if b**2 - 4*a*c < 0:
    print("Phương trình vô nghiệm.")    
elif b**2 - 4*a*c == 0:
    x = -b / (2*a)
    print(f"Phương trình có một nghiệm: x = {x:.2f}")
else:
    x1 = (-b + (b**2 - 4*a*c)**0.5) / (2*a)
    x2 = (-b - (b**2 - 4*a*c)**0.5) / (2*a)
    print(f"Phương trình có hai nghiệm: x1 = {x1:.2f}, x2 = {x2:.2f}")
```

```python
def phuongtrinhbac2(a, b, c):
    if a == 0:
        return phuongtrinhbac1(b, c)
    else:
        dieukien = b**2 - 4*a*c
        if dieukien < 0:
            return "Phương trình vô nghiệm."
        elif dieukien == 0:
            x = -b / (2*a)
            return f"Phương trình có một nghiệm: x = {x:.2f}"
        else:
            x1 = (-b + dieukien**0.5) / (2*a)
            x2 = (-b - dieukien**0.5) / (2*a)
            return f"Phương trình có hai nghiệm: x1 = {x1:.2f}, x2 = {x2:.2f}"
print(phuongtrinhbac2(1, -3, 2))
```

Bài 8: Giải hệ phương trình tuyến tính:
ax  by  m
cx  dy  n

```python
a,b,c,d,m,n = map(float, input("hệ số ma trận: ").split())
dinhthuc = a * d - b * c
dinhthuccotx = m * d - b * n
dicnhthuccoty = a * n - m * c
if dinhthuc != 0:
    x = dinhthuccotx / dinhthuc
    y = dicnhthuccoty / dinhthuc
    print(f"Hệ có nghiệm duy nhất: x = {x}, y = {y}")
else:
    if dinhthuccotx == 0 and dicnhthuccoty == 0:
        print("Hệ có vô số nghiệm")
    else:
        print("Hệ vô nghiệm")
```

```python
def hephuongtrinhhaian(a, b, m, c, d, n):
    dinhthuc = a * d - b * c
    dinhthuccotx = m * d - b * n
    dicnhthuccoty = a * n - m * c

    if dinhthuc != 0:
        x = dinhthuccotx / dinhthuc
        y = dicnhthuccoty / dinhthuc
        return f"Hệ có nghiệm duy nhất: x = {x}, y = {y}"
    else:
        if dinhthuccotx == 0 and dicnhthuccoty == 0:
            return "Hệ có vô số nghiệm"
        else:
            return "Hệ vô nghiệm"
a,b,c,d,m,n = map(float, input("hệ số ma trận: ").split())
print(hephuongtrinhhaian(a, b, m, c, d, n))
```

Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vào một điểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?

```python
# tâm O (x, y); điểm A (x1, y1)
r,x,y,x1,x2 = map(float, input().split())
if (x - x1)**2 + (y - x2)**2 < r**2:
    print("Điểm A nằm trong đường tròn.")
elif (x - x1)**2 + (y - x2)**2 == r**2:
 print("Điểm A nằm trên đường tròn.")    
elif (x - x1)**2 + (y - x2)**2 > r**2:
 print("Điểm A nằm ngoài đường tròn.")
```

```python
def xacdinhvitridiemtrongduongtron(r, x, y, x1, x2):
    if (x - x1)**2 + (y - x2)**2 < r**2:
        return "Điểm A nằm trong đường tròn."
    elif (x - x1)**2 + (y - x2)**2 == r**2:
        return "Điểm A nằm trên đường tròn."    
    else:
        return "Điểm A nằm ngoài đường tròn."
r,x,y,x1,x2 = map(float, input().split())
print(xacdinhvitridiemtrongduongtron(r, x, y, x1, x2))
```
