W4A18 Nhập vào số nguyên dương a và b, in toàn bộ ước chung của a và b 


```python
a,b=map(int,input().split())
uocchung=[]
uoc_a=[]
uoc_b=[]
while a<=0 or b<=0:
    print("nhập lại")
for i in range(1,a+1):
    if a%i==0:
     uoc_a.append(i)
for j in range(1,b+1):
    if b%j==0:
     uoc_b.append(j)
for uoc in uoc_a:
   if uoc in uoc_b:
      uocchung.append(uoc)
print(uocchung)
```

W4A21 Nhập vào số nguyên dương n, tính tổng các chữ số của n 

```python
def timsochuso(n) :
    for i in range(0,n):
        if n ==10**i:
            return i+1
        if n<10**i:
            return i       
n=int(input())
sochuso=timsochuso(n)
tongcacchuso=0
a=n/(10**sochuso)
for i in range(1,sochuso+1):
    b=a*10**i-int(a*10**(i-1))*10
    tongcacchuso=tongcacchuso+int(b)
print(tongcacchuso)
```

W4A26
[Fibonaci1]
Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ
bằng tổng hai số trước đó
Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt
quá A

```python
A = int(input("Nhập số nguyên dương A: "))
a = 1
b = 1
if A < 1:
    print("Không có số Fibonacci nào phù hợp.")
elif A == 1:
    print("Số Fibonacci lớn nhất không vượt quá A là: 1")
else:
    sof = a+b
    while sof <= A:
        a = b
        b = sof
        sof = a+b
    print("Số Fibonacci lớn nhất không vượt quá", A, "là:", b)
```
