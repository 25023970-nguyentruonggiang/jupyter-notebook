Viết một hàm duy nhất smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính và trả về total (tổng tiền phải trả,
kiểu số nguyên, đơn vị VND).
Các quy tắc cần cài đặt:
1. Giá mở cửa (0–0.8 km):
- Xe 4 chỗ: 11,000 VND
- Xe 7 chỗ: 12,000 VND
2. Giá theo km sau 0.8 km:
- Xe 4 chỗ: từ 0.8→20 km ơnh 13,000 VND/km; trên 20 km tính 10,500 VND/km
- Xe 7 chỗ: từ 0.8→20 km ơnh 14,000 VND/km; trên 20 km tính 12,000 VND/km
3. Phí chờ: 5 phút đầu miễn phí; sau đó, mỗi 5 phút (làm tròn lên theo từng block 5 phút) tính 2,000 VND.
4. Phụ thu ban đêm: nếu start_hour thuộc [22,23,0,1,2,3,4,5], cộng +20% trên tiền tạm tính (mở cửa + quãng đường +
chờ).
5. Điều kiện đặc biệt:
- Nếu km > 30 và không phải ban đêm, áp dụng giảm 10% trên tổng tiền.
- Nếu tổng tiền cuối cùng < 50,000 nhưng km > 5, đặt tổng tiền bằng 50,000 VND (mức tối thiểu). 


```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type == 4:
        price1 = 11000
        price2 = 13000
        price3 = 10.5000
    else: 
        price1 = 12000
        price2 = 14000
        price3 = 12000
    if km <= 0.8:
        price4 = price1
    elif km <= 20:
        price4 = price1 + (km - 0.8) * price2
    else:
        price4 = price1 + 19 * price2 + (km - 20) * price3
    price4 = 0
    if wait_minutes > 5:
        price5 = (wait_minutes - 5) * 3000
    all_price = price4 + price5
    if (0 <= start_hour <= 5) or start_hour == 22 or start_hour == 23:
        all_price = all_price * 1.2
    if all_price< 50000 and km>30:
        return "50000 VNĐ"
    return round(all_price)
car_type, start_hour, km, wait_minutes=map(int,input().split())
print(smart_taxi_fare(car_type, start_hour, km, wait_minutes))
```

```python
def goc(a,b,c,d,e,f):
    ba_x, ba_y = a - c, b - d
    bc_x, bc_y = e - c, f - d
    vo_huong= ba_x * bc_x + ba_y * bc_y
    ba = (ba_x**2 + ba_y**2)**0.5
    bc = (bc_x**2 + bc_y**2)**0.5
    if ba == 0 or bc == 0:
        return 0
    cos_abc = vo_huong / (ba * bc)
    goc_abc=(2*(1-cos_abc))**0.5
    return goc_abc
a,b,c,d,e,f=map(int,input().split())
goc(a,b,c,d,e,f)
```

và False nếu ngược lại.
- Đầu vào:
- quad: danh sách hoặc bộ gồm đúng 4 đỉnh [(x1,y1), (x2,y2), (x3,y3), (x4,y4)] theo thứ tự (tứ giác có thể lồi hoặc lõm,
nhưng không tự cắt).
- p: một điểm (x, y).
- Đầu ra: giá trị boolean.
- Gợi ý: Cài đặt phép kiểm tra điểm trong đa giác tổng quát (ví dụ: ray casting), đồng thời tính cả trường hợp điểm nằm
trên biên là ở trong. 


```python
def  point_in_quadrilateral(quad, p):
    def cross_product(o, a, b):
        return (a[0] - o[0]) * (b[1] - o[1]) - (a[1] - o[1]) * (b[0] - o[0])
    px, py = p
    signs = []
    for i in range(4):
        p1 = quad[i]
        p2 = quad[(i + 1) % 4]
        cp = cross_product(p1, p2, p)
        signs.append(cp)
    is_all_non_negative = all(s >= 0 for s in signs)
    is_all_non_positive = all(s <= 0 for s in signs)
    return is_all_non_negative or is_all_non_positive
```

30 Yêu cầu: Viết hàm detect_number_type(s) trả về 'int' nếu s biểu diễn một số nguyên, 'float' nếu biểu diễn một số thực dấu
chấm động, ngược lại trả về 'neither'.
- Đầu vào: chuỗi s.
- Đầu ra: một trong các giá trị {'int', 'float', 'neither'}.
- Quy tắc:
- Cho phép có dấu + hoặc - ở đầu.
- Số nguyên phải chỉ gồm các chữ số và không có dấu chấm thập phân (ví dụ: '100').
- Số thực phải có dấu chấm và có chữ số ở cả hai phía của dấu chấm (ví dụ: '100.0', '-3.14').
- Các chuỗi như '.5', '5.', '1e3' được xem là 'neither' trong bài này. 

```python
def detect_number_type(s):
    temp_s = s
    if temp_s.startswith('+') or temp_s.startswith('-'):
        temp_s = temp_s[1:]
    if not temp_s:
        return 'neither'
    if temp_s.isdigit():
        return 'int'
    if '.' in temp_s:
        parts = temp_s.split('.')
        if len(parts) == 2 and parts[0].isdigit() and parts[1].isdigit():
            return 'float'
    return 'neither'
```

- Đầu ra: bộ hai số nguyên (q, r).
- Ví dụ: divide_with_remainder(5, 3) -> (1, 2). 


```python
def divide_with_remainder(dividend, divisor):
    if divisor == 0:
        return "nhập lại"
    q = dividend // divisor
    r = dividend % divisor
    return q, r
a,b=map(int,input().split())
print(*divide_with_remainder(a, b))
```

37 Viết hàm trả về tam giác vuông cân cao n bằng ký tự ''.

Ví dụ với n = 4 


```python
def ve_tam_giac(n):
    a = []
    for i in range(1, n + 1):
        a.append('"' * i)
    return "\n".join(a)
n=int(input())
print(ve_tam_giac(n))
```

44 Viết hàm trả về tích các số nguyên từ 1 đến n (giai thừa). Với n=0, trả về 1.

```python
def giathua(a):
    i=1
    sum=1
    if a==0:
        return "1"
    while i<=a:
        sum=i*sum
        i+=1
    return sum
a=int(input())
print(giathua(a))
```

45 Viết hàm trả về số lượng số nguyên dương k sao cho 1 <= k < n và k > (n - 1) / 2. (Ví dụ: n=9 -> 4).

```python
def dem_so_k(n):
    if n <= 1:
        return 0
    sok = 0
    min = (n - 1) / 2
    for k in range(1, n):
        if k > min:
         sok += 1   
    return sok
n=int(input())
dem_so_k(n)
```
