"[SumPrimesInRange]
Nhập hai số nguyên dương a, b (a ≤ b). Hãy tính tổng các số nguyên tố trong đoạn
[a, b]. Mỗi dòng là một bộ test (a b)."


```python

def songuyento(x):
    if x < 2:
        return False
    for i in range(2, int(x**0.5) + 1):
        if x % i == 0:
            return False
    return True

while True:
        try:
            a, b = map(int, input().split())
            if a>b:
                print("nhập lại a<=b")
                continue
            tong=0
            for i in range(a,b+1):
                if songuyento(i):
                    tong = tong + i
            print("tổng các số nguyên tố từ",a,"đến",b,"là:",tong)
        except EOFError:
            break
        except:
            break
```

W4A8
[ReverseAddPalindrome]
Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số đối xứng (palindrome). In ra số bước và giá trị palindrome thu được. 

```python
def palindrome(n):
    s = str(n)
    return s == s[::-1]
def reverse_n(n):
    return int(str(n)[::-1])
n = int(input("Nhập n: "))
buoc = 0
while not palindrome(n):
    n = n + reverse_n(n)
    buoc += 1
print(f"Số bước: {buoc}")
print(f"Số đối xứng thu được: {n}")
```

W4A9
[PerfectSquaresDistinctDigits]
Nhập vào số nguyên dương n. In ra tất cả các số chính phương ≤ n có các chữ số
đôi một khác nhau (in theo thứ tự tăng dần, cùng hàng, cách nhau bởi một dấu
cách). Nếu không có, in ra (no number).


```python
def doimot(n):
    s = str(n)
    return len(s) == len(set(s))
def kiemtrachinhphuong(n):
    sochinhphuong= []
    i = 1
    while True:
        square = i * i
        if square > n:
            break
        if doimot(square):
            sochinhphuong.append(str(square))
        i += 1
    if not sochinhphuong:
        print("(no number)")
    else:
        print(" ".join(sochinhphuong))
n = int(input("Nhập n: "))
kiemtrachinhphuong(n)
```

W4A10
"[LongestCollatzUpToN]
Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất
và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số
nhỏ nhất."

```python
def timdodaiday(x):
    dodai= 1
    while x != 1:
        if x % 2 == 0:
            x //= 2
        else:
            x = 3 * x + 1
        dodai += 1
    return dodai
def timdodailonnhat(n):
    dodaimax = 0
    xbenhat = 0 
    for i in range(1, n + 1):
        dodaicuasoi = timdodaiday(i)
        if dodaicuasoi > dodaimax:
            dodaimax = dodaicuasoi
            xbenhat = i
    print(f"{xbenhat} {dodaimax}")
n = int(input())
timdodailonnhat(n)
```

Số này bằng tổng tất cả các ước của số kia (trừ chính nó) và ngược lại.
Ví dụ 220 và 284 là một cặp số thân thiết
220 ngoài chính nó ra, nó còn có 11 ước số khác là: 1, 2, 4, 5, 10, 11, 20, 44, 55,
110. Tổng của 11 ước này cho chúng ta 284
284 ngoài bản thân nó, còn có 5 ước số khác là: 1, 2, 4, 71, 142. Tổng của chúng
vừa đúng bằng 220
Nhập hai số từ bàn phím và kiểm tra xem chúng có phải là cặp số thân thiết không.
Nếu có, in ra màn hình "true", ngược lại, in ra "false"

```python
a,b = map(int, input().split())
i=1
j=1
uoc_a=[]
uoc_b=[]
while i<a:
    if a%i==0:
        uoc_a.append(i)
    i+=1
while j<b:
    if b%j==0:
        uoc_b.append(j)
    j+=1
if sum(uoc_a)==b and sum(uoc_b)==a:
    print("true")
else:
    print("false")
```
